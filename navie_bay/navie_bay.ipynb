```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn import tree
```

```python
df = pd.read_csv('./../patient2025 - patient2025.csv')
df.head()
```

fill BMI na ด้วย median

```python
df["BMI"] = df["BMI"].fillna(df["BMI"].median())
df
```

Fill excrcise na ด้วย 0

```python
df["Exercise (min/week)"] = df["Exercise (min/week)"].fillna(0)
df
```

replace Diabetes yes nno ด้วย 0 กับ 1 

```python
df["Diabetes"] = df["Diabetes"].replace({
    "yes": 1,
    "No": 0,})
df
```

แยกข้อมูลซ้ำดูจาก HN (ไม่มีซ้ำ) 

```python
df = df.drop_duplicates(subset="HN")
df
```

drop collum ที่ไม่ใช้

```python
df = df.drop(columns = ['HN', 'Marriage Status', 'Living Area'])
df
```

replace Chorestorol unknown ด้วย nan

```python
df["Chorestorol"] = (
    df["Chorestorol"]
    .replace("Unknown", np.nan)
    )
df
```

ทำค่าช่วง Chorestorol ให้เป็น mean

```python
df[["chol_low", "chol_high"]] = (
    df["Chorestorol"]
    .str.split("-", expand=True)
)

df[["chol_low", "chol_high"]] = df[["chol_low", "chol_high"]].astype(float)

df["Chorestorol"] = df[["chol_low", "chol_high"]].mean(axis=1)
df.drop(columns=["chol_low", "chol_high"], inplace=True)
df
```

fill na ของ Chorestorol ด้วย median (แปลกๆ) ฝากเช็คที

```python
df["Chorestorol"] = df["Chorestorol"].fillna(df["Chorestorol"].median())
df
```
