```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.naive_bayes import MultinomialNB, GaussianNB, BernoulliNB
```

```python
df = pd.read_csv('./../patient2025 - patient2025.csv')
df.head()
```

fill BMI na ด้วย median

```python
df["BMI"] = df["BMI"].fillna(df["BMI"].median())
df
```

Fill excrcise na ด้วย 0

```python
df["Exercise (min/week)"] = df["Exercise (min/week)"].fillna(0)
df
```

replace Diabetes yes no ด้วย 0 กับ 1 

```python
df["Diabetes"] = df["Diabetes"].replace({
    "yes": 1,
    "No": 0,})
df
```

แยกข้อมูลซ้ำดูจาก HN (ไม่มีซ้ำ) 

```python
df = df.drop_duplicates(subset="HN")
df
```

drop column ที่ไม่ใช้

```python
df = df.drop(columns = ['HN', 'Marriage Status', 'Living Area'])
df
```

replace Chorestorol unknown ด้วย nan

```python
df["Chorestorol"] = (
    df["Chorestorol"]
    .replace("Unknown", np.nan)
    )
df
```

ทำค่าช่วง Chorestorol ให้เป็น mean

```python
df[["chol_low", "chol_high"]] = (
    df["Chorestorol"]
    .str.split("-", expand=True)
)

df[["chol_low", "chol_high"]] = df[["chol_low", "chol_high"]].astype(float)

df["Chorestorol"] = df[["chol_low", "chol_high"]].mean(axis=1)
df.drop(columns=["chol_low", "chol_high"], inplace=True)
df
```

fill na ของ Chorestorol ด้วย median (แปลกๆ) ฝากเช็คที

```python
df["Chorestorol"] = df["Chorestorol"].fillna(df["Chorestorol"].median())
df

# debug = df["Chorestorol"]
# ex_list = []

# for e in debug:
#     if e not in ex_list:
#         ex_list.append(e)

# for ex in ex_list:
#     print(ex)
```

Gaussian Naive Bayes test

```python
features = ['FBS', 'BMI', 'age', 'Chorestorol', 'Diabetes', 'hypertension', 'vegetarian (1= yes, 0=no)']
X = df[features] # features
y = df.stroke # class

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 6)

model = GaussianNB()

model.fit(X_train, y_train)

predict_test = model.predict(X_test)

print(f"Accuracy: {metrics.accuracy_score(y_test, predict_test)}")
print(metrics.classification_report(y_test, predict_test))
```

```python
features = ['Diabetes', 'hypertension', 'vegetarian (1= yes, 0=no)']
X = df[features] # features
y = df.stroke # class

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 6)

model = BernoulliNB()

model.fit(X_train, y_train)

predict_test = model.predict(X_test)

print(f"Accuracy: {metrics.accuracy_score(y_test, predict_test)}")
print(metrics.classification_report(y_test, predict_test))
```

ลองเลือก feature มาบางส่วนเพื่อใช้ในการ train โดยจะเลือกเฉพาะ feature ที่เป็น 0 หรือ 1 ใช้ MultinomialNB เป็น model ในการ train (Current Best)

```python
features = ['Diabetes', 'hypertension', 'vegetarian (1= yes, 0=no)']
X = df[features] # features
y = df.stroke # class

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 6)

model = MultinomialNB()
model.fit(X_train, y_train)

predict_test = model.predict(X_test)

print(f"Accuracy: {metrics.accuracy_score(y_test, predict_test)}")
print(metrics.classification_report(y_test, predict_test))
```
